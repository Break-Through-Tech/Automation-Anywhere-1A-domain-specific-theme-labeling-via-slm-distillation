# Qwen2.5-3B Research Pilot

Quality, latency and cost analysis for the baseline SLM, fine-tuned SLM and Claude teacher model.

## 1. Load the published results

```python
import pandas as pd
import matplotlib.pyplot as plt

BASE_URL = 'https://raw.githubusercontent.com/Break-Through-Tech/Automation-Anywhere-1A-domain-specific-theme-labeling-via-slm-distillation/btt_dk/experiments/qwen25_3b/results/'

summary = pd.read_csv(
    BASE_URL + 'three_model_summary.csv'
)
stats = pd.read_csv(
    BASE_URL + 'cluster_statistics.csv'
)
metrics = pd.read_csv(
    BASE_URL + 'anonymized_prediction_metrics.csv'
)

display(summary.round(4))
```

## 2. Composite quality comparison

```python
ax = summary.plot.bar(
    x='Model',
    y='Composite',
    figsize=(9, 4),
    legend=False
)
ax.set_ylim(0, 5)
ax.set_ylabel('Composite quality (out of 5)')
ax.set_xlabel('')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 3. Label quality metrics

```python
display(summary[[
    'Model',
    'Faithfulness',
    'Relevance',
    'Specificity',
    'Composite',
    'Cosine Score',
    'ROUGE',
    'BertScore'
]].round(4))
```

## 4. Latency and cost comparison

```python
display(summary[[
    'Model',
    'Latency',
    'Cost per 1000',
    'Word compliance %'
]].round(4))

print('Note: Qwen and Claude latency/cost '
      'use different measurement methods.')
```

## 5. Cluster-level statistics

```python
selected = stats[
    stats['Metric'].isin([
        'Composite',
        'Cosine Score',
        'ROUGE',
        'BertScore'
    ])
]

display(selected.round(4))
```

## 6. Research interpretation

Fine-tuning improved the measured average quality metrics, but the four independent ticket clusters are too few to establish statistical significance.

These results are exploratory and should be validated on a larger held-out dataset.
